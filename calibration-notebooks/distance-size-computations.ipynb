```python
import numpy as np
import matplotlib.pyplot as plt

# matplotlib inline
plt.rcdefaults()
plt.rcParams.update({'xtick.labelsize': 14, 'ytick.labelsize': 14})
plt.rcParams.update({'axes.labelsize': 16})
plt.rcParams.update({'legend.fontsize': 12})
plt.rcParams.update({'axes.titlesize': 18})
plt.rcParams.update({'figure.titlesize': 22})

# Update rcParams to set spines to be false for up and right
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False

colors = ['orange', '#35b779', '#31688e', '#440154', 'gray']
names = ['1', '2', '3', '4', 'control']
```

```python
def angle_to_pos(angle, h):
    return h/np.tan(np.radians(angle))
def pos_to_angle(pos, h):
    return np.degrees(np.arctan(h/pos))

y = [2, 3, 4, 5]        # radius, not diameter
target_size = 8# this is for half the stimul

x1 = np.linspace(4, 30, 100)
x2 = np.linspace(9.5, 39.5, 100)
x = np.linspace(9.5, 30, 100)

# set the max of the minims of x1 and x2
min_x = max(min(x1), min(x2))
max_x = min(max(x1), max(x2))

print('X range is from', min_x, 'to', max_x)
print('Position for stim 1 is at', angle_to_pos(target_size, y[0]), 'cm')
print('Position for stim 2 is at', angle_to_pos(target_size, y[1]), 'cm')
print('Position for stim 3 is at', angle_to_pos(target_size, y[2]), 'cm')
print('Position for stim 4 is at', angle_to_pos(target_size, y[3]), 'cm')

print('Displacement is from', angle_to_pos(target_size, y[1])-angle_to_pos(target_size, y[0]), 'cm')

alpha_1 = pos_to_angle(x, y[0])
alpha_2 = pos_to_angle(x, y[1])
alpha_3 = pos_to_angle(x, y[2])
alpha_4 = pos_to_angle(x, y[3])

plt.figure()
plt.plot(x, alpha_1, color='orange', label=str(y[0]) + ' cm')
plt.plot(x, alpha_2, color='#35b779', label=str(y[1]) + ' cm')
plt.plot(x, alpha_3, color='#31688e', label=str(y[2]) + ' cm')
plt.plot(x, alpha_4, color='#440154', label=str(y[3]) + ' cm')

plt.scatter(angle_to_pos(target_size, y[0]), target_size, color='orange')
plt.scatter(angle_to_pos(target_size, y[1]), target_size, color='#35b779')
plt.scatter(angle_to_pos(target_size, y[2]), target_size, color='#31688e')
plt.scatter(angle_to_pos(target_size, y[3]), target_size, color='#440154')

plt.axhline(y=target_size, color='gray', linestyle='--')
plt.axvline(min_x, color='gray', linestyle='--')
plt.axvline(max_x, color='gray', linestyle='--')
plt.xlabel('Distance (cm)')
plt.ylabel('Visual angle (degrees)')
plt.legend()
plt.title('Visual angle as a function of distance')

# check the difference in visual angle between start at 9.5 and finish at 30
difference_5 = pos_to_angle(30, 5) - pos_to_angle(9.5, 5)
difference_4 = pos_to_angle(30, 4) - pos_to_angle(9.5, 4)
difference_3 = pos_to_angle(30, 3) - pos_to_angle(9.5, 3)
difference_2 = pos_to_angle(30, 2) - pos_to_angle(9.5, 2)

# print results
print('Difference in visual angle for 5 cm:', difference_5)
print('Difference in visual angle for 4 cm:', difference_4)
print('Difference in visual angle for 3 cm:', difference_3)
print('Difference in visual angle for 2 cm:', difference_2)
```
