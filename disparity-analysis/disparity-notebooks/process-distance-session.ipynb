# Process the distance (radial) session

Builds the `RadialDevice`, runs `SessionAnalyzer.analyze_session_stats`, and saves
`{session_name}_dataframe.pkl` next to the session data. That pkl is what
`monocular-disparity-distance-comparison.ipynb` reads for this experiment via
`CrossExperimentAnalysis.from_devices(R, M, D)`.

Run this in its own kernel. It puts **only** `radial-analysis` on `sys.path`:
`radial-analysis` and `disparity-analysis` both define a `src.analysis.CellAnalysis`,
so mixing them in one kernel would import the wrong one.

```python
import sys, os
import pandas as pd

base_dir = r"C:\Users\acostamendoza\projects\distance-project"
for p in (base_dir, base_dir + r"\radial-analysis"):
    if p not in sys.path:
        sys.path.append(p)

from src.data.RadialDevice import RadialDevice
from src.analysis.CellAnalysis import SessionAnalyzer

%load_ext autoreload
%autoreload 2
```

```python
mouse, date, experiment = "SAM045", "260721", "distance"
R = RadialDevice(mouse_name=mouse, record_date=date, experiment_name=experiment)
print("session:", R.session_name, "| cells:", len(R.suite2p_id))
```

```python
# load cached stats if present, else compute once (slow) and save the pkl the
# comparison notebook reads
pkl_path = os.path.join(R.data_dir, f"{R.session_name}_dataframe.pkl")
if os.path.exists(pkl_path):
    stats_df = pd.read_pickle(pkl_path)
    print(f"Loaded cached stats: {pkl_path}")
else:
    stats_df = SessionAnalyzer(R).analyze_session_stats(n_iterations=200)
    SessionAnalyzer(R).save_dataframe(stats_df)   # -> pkl_path
stats_df.head()
```
