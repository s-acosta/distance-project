```python
def compute_session(mouse_name, record_date, experiment_name, method='ANOVA'):

    data_dir = r'D:/data-distance-coding-project/processed-data/{record_date}_{mouse_name}_{experiment_name}/'.format(record_date=record_date, mouse_name=mouse_name, experiment_name=experiment_name)
    
    if not os.path.exists(data_dir):
        raise FileNotFoundError(f"The directory {data_dir} does not exist.")
    else:
        print(f"The directory {data_dir} exists.")
        results_path = create_new_result_dir(data_dir, r'results')

    R = RadialDevice(data_dir, response_mode='spikes')
    Plots = RadialPlots(R)
    stats = R.analyze_session_stats(n_iterations=100, method=method)
    
    Summ = RadialSummaries(R, stats, Plots)

    return Summ

def save_summ(Summ, summ_file):

    with open(summ_file, 'wb') as f:
        pickle.dump(Summ, f)
   
```

```python
for mouse_name, record_date, experiment_name in zip(mouse_names, record_dates, experiment_names):

    base_dir = r'D:/data-distance-coding-project/processed-data/' + mouse_name + '_radial'
    Summ = compute_session(mouse_name, record_date, experiment_name, method='ANOVA')
    
    summ_file = os.path.join(base_dir, f"{record_date}_{mouse_name}_{experiment_name}_summ.pkl")
    save_summ(Summ, summ_file)
```
