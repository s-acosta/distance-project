## Probability map: single trial prediction

```python
from src.decoding.NNDecoder import JointDecoder

# Get data
agg = SessionAggregator('SAM040', 'tiling_120um')
X, y_distance = agg.get_decoder_data(monocular=0, area='V1')

# Manual 80/20 stratified split
n_distances = 4
trials_by_dist = {d: np.where(y_distance == d)[0] for d in range(n_distances)}

train_idx, test_idx_by_dist = [], {}
for dist in range(n_distances):
    idxs = np.random.permutation(trials_by_dist[dist])
    split = int(0.8 * len(idxs))
    train_idx.extend(idxs[:split])
    test_idx_by_dist[dist] = idxs[split:]

# Train joint decoder
joint_decoder = JointDecoder()
joint_decoder.fit(X[:, :, train_idx], y_distance[train_idx])

# Pick a held-out trial (distance=2)
test_dist = 2
held_out_idx = test_idx_by_dist[test_dist][0]
trial_response = X[:, :, held_out_idx]  # (n_neurons, n_angles)

# Static plot for one time-point
angle_idx = 30
response_vector = trial_response[:, angle_idx]
angle_range = np.linspace(-30, 60, X.shape[1])
fig, joint, p_dist, p_angle = joint_decoder.plot_joint(
    response_vector,
    angle_values_deg=angle_range,
    actual_distance=test_dist,
    actual_angle_idx=angle_idx,
)

# Animate the full trial
joint_decoder.animate_trial(
    trial_response,
    angle_values_deg=angle_range,
    actual_distance=test_dist,
    save_path='joint_decoder_trial_dist2.gif',
    fps=5,
)
```
