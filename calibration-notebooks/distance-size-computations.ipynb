```python
import numpy as np
import matplotlib.pyplot as plt

# matplotlib inline
plt.rcdefaults()
plt.rcParams.update({'xtick.labelsize': 14, 'ytick.labelsize': 14})
plt.rcParams.update({'axes.labelsize': 16})
plt.rcParams.update({'legend.fontsize': 12})
plt.rcParams.update({'axes.titlesize': 18})
plt.rcParams.update({'figure.titlesize': 22})

# Update rcParams to set spines to be false for up and right
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False

colors = ['orange', '#35b779', '#31688e', '#440154', 'gray']
names = ['1', '2', '3', '4', 'control']
```

```python
def angle_to_pos(angle, h):
    return h/np.tan(np.radians(angle))
def pos_to_angle(pos, h):
    return np.degrees(np.arctan(h/pos))

y = [2, 4]        # radius, not diameter
target_size = 10     # this is for half the stimuli
delta_size = 4

x = np.linspace(1, 41, 100)

plt.figure()

target_position = []
alpha = []
for i in range(len(y)):
    alpha.append(pos_to_angle(x, y[i]))
    target_position.append(angle_to_pos(target_size, y[i]))
    plt.plot(x, alpha[i], color=colors[i], label=str(y[i]) + ' cm')
    
    plt.scatter(angle_to_pos(target_size + delta_size, y[i]), target_size + delta_size, color=colors[i])
    print('Starting position and angle for distance', y[i], 'cm:', angle_to_pos(target_size + delta_size, y[i]), target_size + delta_size, 'degrees')
    print('Ending position and angle for distance', y[i], 'cm:', angle_to_pos(target_size - delta_size, y[i]), target_size - delta_size, 'degrees')
    plt.scatter(angle_to_pos(target_size - delta_size, y[i]), target_size - delta_size, color=colors[i])
    
# start of distance one
plt.axhline(y=target_size+delta_size, color='gray', linestyle='--')
plt.axhline(y=target_size-delta_size, color='gray', linestyle='--')
plt.xlabel('Distance (cm)', fontsize=16)
plt.ylabel('Visual angle (degrees)', fontsize=16)
plt.legend()
plt.title('Visual angle as a function of distance')
```
